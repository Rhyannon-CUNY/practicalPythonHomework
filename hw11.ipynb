```python
import pandas as pd
import regex as re
```

```python
with open("confession-judgments.txt", "r") as text_obj:
    text = text_obj.read()
```

```python
date_received_pat = re.compile(r"NYSCEF: (\d{1,2}/\d{1,2}/\d{4})", re.M)
date_received = date_received_pat.findall(text)
date_received
```

```python
case_no_pat = re.compile(r"INDEX NO. (\d+/\d{4})", re.M)
case_no = case_no_pat.findall(text)
case_no
```

```python
repay_amt_pat = re.compile(r"\(\$(\d+.\d+)\)?. within", re.M)
repay_amt = repay_amt_pat.findall(text)
repay_amt
```

```python
repay_date_pat = re.compile(r"month by (\w+\s\d+,\s\d{4})", re.M)
repay_date = repay_date_pat.findall(text)
repay_date
```

```python
## for loop to capture all target data

df_list = []
case_no_pat = re.compile(r"INDEX NO. (\d+/\d{4})", re.M)
date_received_pat = re.compile(r"NYSCEF: (\d{1,2}/\d{1,2}/\d{4})", re.M)
repay_amt_pat = re.compile(r"\(\$(\d+.\d+)\)?. within", re.M)
repay_date_pat = re.compile(r"month by (\w+\s\d+,\s\d{4})", re.M)

with open("confession-judgments.txt", "r") as text_obj:
    text = text_obj.read()
    target_list = [case_no_pat, 
                   date_received_pat, 
                   repay_amt_pat, 
                   repay_date_pat
                  ]
    for i, target in enumerate(target_list, start =0):
        print(f"processing {i} of {len(target_list)}...")
        data = target.findall(text)
        df_list.append(data)

df_list
```

```python
columns = ["case_no", "date_received", "repay_amt", "repay_date"]

df = pd.DataFrame(zip(*df_list), columns=columns)
df
```
